# 0.  Load the darkplotter library

```python
# import darkplotter lib
import darkplotter as dp
```

# 1. Call the Data manager constructor

the constructor automatically loads the content of the json folder where all available data are present. You can specify as well your own folder with the 'path' flag

```python
mydata = dp.DMdata(path='./json')
```

# 2. Select the experiment you want
By default, all data are shown. By setting "collaboration" and "experiment" flags you can restrict the selection to specific elements

```python
import pandas as pd

myexp = pd.concat([
    mydata.get_experiment(label=[
        "LZ 2025",
        "XENON1T S2-only Migdal",
        "DarkSide-50 Migdal 2022",
        "XENONnT 2025",
        #"PandaX-4T 2025",
        "Xe_SI",
        "PandaX S2-only Migdal",
        "LZ Migdal",
        "XENONnT 2026 S2-only",
        "PandaX-4T 2025 S2-only",
#        "XENONnT 2025 Neutrino Fog",
        "DarkSide-50 2025",
        "LZ 2025 Neutrino Fog",
        #"XENONnT/LZ",
        #"DarkSide20k 200ty",
        #"ARGO 1000ty",
        #"PandaX-xT 200ty",
        #"XLZD 200ty",
        #"XLZD 1000ty"
    ])
    ])
```

# 4. Call the DMplotter constructor 
## Change the massunit by setting "massunit"

```python
myplotter = dp.DMplotter()
#myplotter.figlimits = {'xmin':5, 'xmax':1e3, 'ymin':1e-50, 'ymax':1e-41}
#myplotter.figlimits = {'xmin':1, 'xmax':1e3, 'ymin':1e-50, 'ymax':1e-41}
#myplotter.figlimits = {'xmin':1, 'xmax':1e1, 'ymin':1e-48, 'ymax':1e-40}
myplotter.figlimits = {'xmin':0.1, 'xmax':1e3, 'ymin':1e-50, 'ymax':1e-34}

myplotter.plot(myexp,massunit="GeV")
```
